# 🛡️ Hash de Senhas com Argon2

Este notebook demonstra como usar o algoritmo **Argon2** para gerar e verificar hashes de senhas de forma segura, seguindo as melhores práticas de segurança.

## 📋 Pré-requisitos
- Execução no Google Colab (ou Python 3.11+ com `argon2-cffi` instalado).
- Pacotes necessários: `argon2-cffi`.

## 🔧 Instalação
```bash
pip install argon2-cffi
```

## ⚙️ Configuração
Os parâmetros do Argon2 estão definidos no dicionário `config`:
```python
config = {
    "salt": os.urandom(16),  # Gerado automaticamente (não utilizado diretamente pelo PasswordHasher)
    "time_cost": 2,          # Número de iterações (aumente para maior segurança)
    "memory_cost": 65536,    # Memória em KiB (64 MiB)
    "parallelism": 4,        # Número de threads
    "hash_len": 32,          # Tamanho do hash em bytes (256 bits)
}
```

## 🚀 Funcionalidades
### 1. Geração de Hash
- Usa `PasswordHasher` para criar um hash seguro da senha "senha123".
- O hash inclui automaticamente:
  - Salt aleatório.
  - Parâmetros usados (custo de tempo, memória, paralelismo).
  - Hash final codificado.

Exemplo de código:
```python
pass_hasher = PasswordHasher(time_cost=2, memory_cost=65536, parallelism=4, hash_len=32)
hash = pass_hasher.hash("senha123")
```

### 2. Verificação de Hash
- Valida se uma senha corresponde ao hash armazenado.
- Lança `VerifyMismatchError` em caso de falha.

Exemplo:
```python
try:
    pass_hasher.verify(hash, "senha123")
    print("Hash válido!")
except VerifyMismatchError:
    print("Hash inválido!")
```

## 📝 Notas Importantes
1. **Salt Automático**: O `PasswordHasher` gera e gerencia o salt internamente (não use o `salt` do dicionário `config`!).
2. **Parâmetros**:
   - Aumente `time_cost` e `memory_cost` para maior segurança (equilíbrio com desempenho).
   - Valores padrão do OWASP: `time_cost=2`, `memory_cost=64 MiB`.
3. **Armazenamento**: Guarde o hash completo (inclui salt e parâmetros) para futuras verificações.

## 🔒 Segurança
- **Argon2id**: Variante padrão do `PasswordHasher` (resistente a side-channel e GPU/ASIC).
- **Memory-hard**: Dificulta ataques paralelizados em hardware especializado.
- **Nunca armazene senhas em texto claro**.

````python
from argon2 import PasswordHasher
import argon2
import os
from argon2.exceptions import VerifyMismatchError

# Configuração
# Sugestão: Colocar as configurações em um arquivo .env
config = {
    "salt": os.urandom(16),  # Salt aleatório
    "time_cost": 2,          # Iterações (t)
    "memory_cost": 65536,    # Memória em KiB (64 MB)
    "parallelism": 4,        # Threads (p)
    "hash_len": 32,          # Tamanho do hash (bytes)
}

# Criando o Hash
pass_hasher = PasswordHasher(
    time_cost=config["time_cost"],
    memory_cost=config["memory_cost"],
    parallelism=config["parallelism"],
    hash_len=config["hash_len"],
)

# Gerando o Hash
hash = pass_hasher.hash("senha123")

# Exibindo o Hash
print(f"Hash gerado: {hash}")
print()

# Verificando o Hash
try:
    pass_hasher.verify(hash, "senha123")
    print("Hash válido!")
except VerifyMismatchError:
    print("Hash inválido!")
````
